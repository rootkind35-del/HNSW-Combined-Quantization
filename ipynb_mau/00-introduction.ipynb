```python
import pandas as pd

df = pd.read_csv("data/flights.csv")

df.head()
```

```python
print("Số dòng:", df.shape[0])
print("Số cột:", df.shape[1])
```

```python
df.info()
```

```python
important_cols = [
    "FL_DATE",
    "OP_UNIQUE_CARRIER",
    "ORIGIN",
    "DEST",
    "DEP_DELAY",
    "ARR_DELAY",
    "CANCELLED",
    "DIVERTED",
    "DISTANCE",
    "CARRIER_DELAY",
    "WEATHER_DELAY",
    "NAS_DELAY",
    "SECURITY_DELAY",
    "LATE_AIRCRAFT_DELAY"
]

available_cols = [col for col in important_cols if col in df.columns]

print(available_cols)
```

```python
df[available_cols].head()
```

```python
df[available_cols].isnull().sum()
```

```python
print("Số dòng bị trùng:", df.duplicated().sum())
```

```python
cancelled_count = (df["arr_cancelled"] == 1).sum()

cancelled_rate = cancelled_count / len(df) * 100

print("Số chuyến bay bị hủy:", cancelled_count)
print("Tỷ lệ chuyến bay bị hủy:", round(cancelled_rate, 2), "%")
```

```python
delay_count = df["arr_del15"].sum()

delay_rate = delay_count / df["arr_flights"].sum() * 100

print("Số chuyến bay đến trễ:", delay_count)
print("Tỷ lệ chuyến bay đến trễ:", f"{round(delay_rate, 2)}%")
```

```python
# Tính số phút trễ khi đến trung bình trên mỗi chuyến bay
avg_arr_delay = df["arr_delay"].sum() / df["arr_flights"].sum()

print("Trễ khi đến trung bình (trên tổng số chuyến bay):", f"{round(avg_arr_delay, 2)} phút")

# Nếu muốn tính trễ trung bình chỉ tính riêng các chuyến bị trễ (arr_del15):
avg_delay_per_delayed_flight = df["arr_delay"].sum() / df["arr_del15"].sum()
print("Trễ khi đến trung bình (chỉ tính chuyến bị trễ):", f"{round(avg_delay_per_delayed_flight, 2)} phút")
```

```python
airline_delay = (
    df.groupby("carrier_name")["arr_delay"]
    .sum()
    .sort_values(ascending=False)
)

print(airline_delay.head(10))
```

```python
top10_airline_delay = airline_delay.head(10)

top10_airline_delay
```

### Average Arrival Delay by Airline

```python
import matplotlib.pyplot as plt

top10_airline_delay.plot(kind="bar", figsize=(10, 5))

plt.title("Top 10 Airlines by Average Arrival Delay")
plt.xlabel("Airline")
plt.ylabel("Average Arrival Delay (minutes)")
plt.xticks(rotation=45)
plt.show()
```

```python
origin_delay = (
    df.groupby("airport_name")["arr_delay"]
    .sum()
    .sort_values(ascending=False)
)

print(origin_delay.head(10))
```

### Average Departure Delay by Origin Airport

```python
top10_origin_delay = origin_delay.head(10)

top10_origin_delay.plot(kind="bar", figsize=(10, 5))

plt.title("Top 10 Origin Airports by Average Departure Delay")
plt.xlabel("Origin Airport")
plt.ylabel("Average Departure Delay (minutes)")
plt.xticks(rotation=45)
plt.show()
```

```python
delay_causes = df[
    [
        "carrier_delay",
        "weather_delay",
        "nas_delay",
        "security_delay",
        "late_aircraft_delay"
    ]
].sum().sort_values(ascending=False)

print(delay_causes)
```

### Analysis of Delay Causes

```python
delay_causes.plot(kind="bar", figsize=(10, 5))

plt.title("Average Delay by Cause")
plt.xlabel("Delay Cause")
plt.ylabel("Average Delay (minutes)")
plt.xticks(rotation=45)
plt.show()
```

## Conclusion

From the January 2008 Airline On-Time Performance dataset:

- The dataset contains 605,765 flight records and 109 attributes.
- There are no duplicated records in the dataset.
- 279,427 flights arrived late.
- The percentage of delayed flights is approximately 47.59%.
- The average departure delay is 11.48 minutes.
- The average arrival delay is 10.19 minutes.
- Delay causes have different impacts on flight performance.
- Carrier-related delays have the highest average delay time among the analyzed delay causes.

Overall, flight delays were relatively common in January 2008. Nearly half of the valid flights arrived later than scheduled. Carrier-related factors, late aircraft, and aviation system factors were important contributors to delays.
